3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = df.pivot_table(index='Branch', columns='Product line', values='Invoice ID', aggfunc='count')
invoices_by_category
```

2. Які категорії товарів мають найменшу кількість покупок у середньому по всіх філіях?

3.Чи витрачають учасники програми лояльності (Member) у середньому більше грошей за одну покупку, ніж звичайні клієнти (Normal)?

```python
avg_total_by_customer = df.groupby("Customer type")["Total"].mean()

avg_total_by_customer
```

Середній чек учасників програми лояльності вищий, що свідчить про ефективність програми. Тоді бізнесу варто заохочувати клієнтів приєднуватися до програми лояльності, наприклад за допомогою бонусів, персональних знижок або спеціальних пропозицій.
